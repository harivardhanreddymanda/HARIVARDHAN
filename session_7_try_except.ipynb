**try-except**

- try except is used to handle the exception mechanism

- try except is used to catch and handle errors

- when we get the error python will stop the exectuition 

- using try and execpt eventhouht we get errors, the execution will not stop

- it will catch all the logical errors automatically

- the actual code will write in **try block**

- if any errors in try block, will catch the errors in **except** block

- syntax errors will not catch
