- Any loop we required 3 things

    - initial point(start point)

    - increment or decrement

    - condition to stop

- in for loop the above three are happens in a single line

- but in while loop all 3 needs 3 different lines

```python
i=1
n=int(input("enetr a number:"))
while i<n:
    print(i)
    i=i+1
```
